# Notebook 01: Carga y Exploración Inicial del Dataset Real (Kaggle Credit Risk)

**Objetivo:** Cargar el dataset real de riesgo crediticio de Kaggle, seleccionar las variables financieras clave definidas para la red neuronal y analizar la estructura inicial y balance de la variable objetivo (`loan_status`).

```python
import os
import pandas as pd
import numpy as np

# URL pública de respaldo del dataset "Credit Risk" de Kaggle
url_dataset = "https://raw.githubusercontent.com/pr1m8/Credit-Default/main/credit_risk_dataset.csv"

# Carga de los datos crudos
df_raw = pd.read_csv(url_dataset)

# Seleccionar las 6 variables explicativas financieras + 1 variable target
cols_interes = [
    'person_age',
    'person_income',
    'loan_amnt',
    'loan_int_rate',
    'loan_percent_income',
    'cb_person_cred_hist_length',
    'loan_status' # Target: 1 = Default / Mora, 0 = Cumple
]

df = df_raw[cols_interes].copy()

print(f"Dataset cargado con éxito. Dimensiones: {df.shape[0]} filas x {df.shape[1]} columnas.")
df.head()
```

```python
# 1. Resumen de tipos de datos y valores no nulos
print("--- Resumen Técnico ---")
df.info()

# 2. Estadísticas descriptivas de las variables
print("\n--- Estadísticas Descriptivas ---")
display(df.describe())

# 3. Distribución porcentual de la variable target (Mora / Default)
print("\n--- Distribución de la Variable Target (loan_status) ---")
print(df['loan_status'].value_counts(normalize=True) * 100)
```

```python
# Crear carpeta para guardar los datos crudos
os.makedirs('data/raw', exist_ok=True)

# Guardar el archivo en la ruta local de ejecución
df.to_csv('data/raw/credit_risk_dataset.csv', index=False)
print("Copia local del dataset guardada en: data/raw/credit_risk_dataset.csv")
```
