# Notebook 04: Arquitectura de la Red Neuronal (Perceptrón Multicapa - MLP)

**Objetivo:** Declarar formalmente la estructura de la red neuronal en PyTorch, definiendo la capa de entrada, las capas ocultas con activación ReLU, y la capa de salida con activación Sigmoid (Sección 6 de la plantilla).

```python
import torch
import torch.nn as nn

# Definición de la Red Neuronal Densa (Perceptrón Multicapa - MLP)
class CreditRiskMLP(nn.Module):
    def __init__(self, input_dim):
        super(CreditRiskMLP, self).__init__()
        # Capa Oculta 1: 6 entradas -> 16 neuronas
        self.hidden1 = nn.Linear(input_dim, 16)
        self.relu1 = nn.ReLU()

        # Capa Oculta 2: 16 neuronas -> 8 neuronas
        self.hidden2 = nn.Linear(16, 8)
        self.relu2 = nn.ReLU()

        # Capa de Salida: 8 neuronas -> 1 neurona (Probabilidad de mora)
        self.output = nn.Linear(8, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        out = self.hidden1(x)
        out = self.relu1(out)
        out = self.hidden2(out)
        out = self.relu2(out)
        out = self.output(out)
        out = self.sigmoid(out)
        return out

# Instanciar el modelo con las 6 variables de entrada de nuestro dataset
model = CreditRiskMLP(input_dim=6)
print("--- Arquitectura del Modelo MLP Instanciada ---")
print(model)
```
