```python
import os
import sys
import torch

# 1. Clonar el repositorio en Colab para obtener la carpeta src oficial (si no existe)
if not os.path.exists('redes-neuronales-riesgo-crediticio'):
    !git clone https://github.com/JuanCardona2203/redes-neuronales-riesgo-crediticio.git

# 2. Agregar la ruta del repositorio al sistema para reconocer la carpeta src
repo_path = '/content/redes-neuronales-riesgo-crediticio'
if repo_path not in sys.path:
    sys.path.append(repo_path)

# 3. Importar la clase directamente desde src/model.py (Cero duplicación)
from src.model import CreditRiskMLP

# 4. Instanciar el modelo
model = CreditRiskMLP(input_dim=6)

# 5. Simular un batch de prueba con 32 muestras y 6 características (Batch size B = 32)
batch_size = 32
input_features = 6

dummy_tensor = torch.randn(batch_size, input_features)
print(f"Dimensiones del tensor de entrada (Batch): {dummy_tensor.shape}")

# 6. Realizar el Forward Pass de prueba
model.eval() # Modo evaluación
with torch.no_grad():
    y_pred = model(dummy_tensor)

print(f"Dimensiones del tensor de salida (Predicciones): {y_pred.shape}")
print("Verificación dimensional exitosa: Las dimensiones de entrada y salida coinciden perfectamente con la teoría.")
```
