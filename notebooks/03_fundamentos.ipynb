# Notebook 03: Fundamentos Matemáticos de la Neurona Artificial

**Objetivo:** Demostrar el cálculo manual y vectorial de una neurona individual (Forward Pass primario) utilizando una muestra real del conjunto de datos procesado (`X_train_scaled`), aplicando la combinación lineal y la función de activación Sigmoid.

```python
import os
import pandas as pd
import numpy as np

# Cargar una fila real del conjunto procesado
url_xtrain = "https://raw.githubusercontent.com/pr1m8/Credit-Default/main/credit_risk_dataset.csv"
# Leemos los datos limpios o tomamos la primera fila como ejemplo de vector de entradas x
df_sample = pd.read_csv(url_xtrain).dropna().head(1)

# Vector de entradas (x) para las 6 variables cuantitativas
cols = ['person_age', 'person_income', 'loan_amnt', 'loan_int_rate', 'loan_percent_income', 'cb_person_cred_hist_length']
x = df_sample[cols].values.flatten()

# Normalización rápida MinMax simulada para la muestra x
x_scaled = (x - x.min()) / (x.max() - x.min() + 1e-8)

print("Vector de entradas normalizado (x):")
for col, val in zip(cols, x_scaled):
    print(f"  - {col}: {val:.4f}")
```

```python
# Definir pesos aleatorios (w) y un bias inicial (b) para la demostración
np.random.seed(42)
w = np.random.uniform(low=-1.0, high=1.0, size=len(x_scaled))
b = 0.05

# 1. Producto punto (Combinación lineal)
z = np.dot(x_scaled, w) + b

print("--- Cálculo Paso a Paso de la Neurona ---")
for i, (col, xi, wi) in enumerate(zip(cols, x_scaled, w)):
    print(f"x_{i+1} ({col[:10]}...): {xi:.4f} * w_{i+1}: {wi:.4f} = {xi*wi:.4f}")

print(f"\nSuma ponderada de productos: {np.dot(x_scaled, w):.4f}")
print(f"Sesgo (b): {b:.4f}")
print(f"Valor lineal (z): {z:.4f}")
```

```python
# Definición de la función Sigmoid
def sigmoid(z_val):
    return 1 / (1 + np.exp(-z_val))

# Activación
a = sigmoid(z)

print("--- Transformación No Lineal (Activación) ---")
print(f"z calculado = {z:.4f}")
print(f"a = Sigmoid(z) = 1 / (1 + e^-({z:.4f}))")
print(f"Salida de la neurona (a / prob): {a:.4f}")
```
